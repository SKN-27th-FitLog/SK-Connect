## GPT-5 이전 모델
### 창작용(Creative Writing)

```python
from langchain_openai import ChatOpenAI

llm_creative = ChatOpenAI(
    model = "gpt-4o-mini",    # 모델 명
    temperature = 0.9,        # 생성 결과의 무작위성 정도 조절 : 값이 낮으면 결정적+안정적/값이 높으면 다양+창의적
    top_p = 0.9,              # 누적확률 컷오프 : 값이 낮으면 가능한 단어 후보를 좁힘
    frequency_penalty = 0.6,  # 반복 단어의 사용빈도에 따른 패널티 조절 : 같은 단어 반복을 줄이고 싶을때 사용
    presence_panalty = 0.6,   # 이미 등장한 단어나 주제를 또 사용할 확률 감소 조절 : 값이 클수록 새로운 단어나 주제로 넘어갈 가능성 커짐
    max_tokens = 500          # 최대 출력 토근 수 제한 : 입력+출력의 토큰 합은 모델의 최대 context 수를 넘을 수 없음
)
```

```python
prompt = "한국어로 희망적인 시를 작성해줘"
# 모델에게 전달할 프롬프트
response = llm_creative.invoke(prompt)
# .invoke() : 모델에 프롬프트를 전달하고 응답 객체를 반환받음
print(response.content)
# .content : 응답 객체에서 실제 텍스트 내용만 출력
```

### 정확한 답변용(Fact-based)

```python
from langchain_openai import ChatOpenAI

llm_precise = ChatOpenAI(
    model= "gpt-4o-mini",       # 사용할 모델을 지정
    temperature=0.1,            # 생성 결과의 무작위성 정도 조절 : 값이 낮으면 결정적이고 안정적인 답
    top_p = 1.0,                # 누적 확률 컷 오프 : 값을 낮추면 가능한 단어 후보 좁힘
    frequency_penalty = 0.0,    # 반복 단어의 사용 빈도에 따른 패널티 조절 : 같은 단어가 반복 괴는 것을 줄이고 싶을 때 사용
    presence_penalty = 0.0,     # 이미 등장한 단어나 주제를 또 사용할 확률 감소 조절 : 값이 클수록 새로운 단어나 주제로 넘어갈 가능성이 커짐 
    max_tokens=300              # 최대 출력 토큰 수 제한
)
```

```python
response = llm_precise.invoke("지구에서 가장 깊은 바다는 어디야?")
# 모델에게 프롬프트 제공
print(response.content)
# 모델의 답병중 content부분만 출력
```

## GPT-5 모델
### 창작용(Creative Writing)

```python
from langchain_openai import ChatOpenAI

creative_llm = ChatOpenAI(
    model = "gpt-5-nano",           # 모델 지정
    reasoning_effort = "medium"     # 복잡한 작업에 high, 단순 요청엔 minimal 권장
)
```

```python
prompt = "한국어로 사람들에게 힘을 주는 시를 하나 만들어줘"

response = creative_llm.invoke(prompt)
# 생성된 모델 객체에 프롬프트를 제공후 받은 답변 저장

print(response.content)
# 모델 객체를 통해 받은 답변 중 답에 대한 답만 출력
```

### 정확한 답변용(Fact-based)

```python
from langchain_openai import ChatOpenAI

creative_llm = ChatOpenAI(
    model = "gpt-5-nano",           # 모델 지정
    reasoning_effort = "high"       # 복잡한 작업에 high -> 논리성 강화
)
```

```python
prompt = "지구에서 가장 깊은 바다는 어디야?"

response = creative_llm.invoke(prompt)
# 생성된 모델 객체에 프롬프트를 제공후 받은 답변 저장

print(response.content)
# 모델 객체를 통해 받은 답변 중 답에 대한 답만 출력
```
