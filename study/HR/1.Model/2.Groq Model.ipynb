### 창작용(Creative Writing)

```python
from langchain_groq import ChatGroq

# 1. 창의적 모델(Creative Model)
# - temperature 높음(1.2) : 다양하고 창의적인 답변
creative_llm = ChatGroq(
    model = "openai/gpt-oss-120b",      # 모델을 지정
    temperature=1.2,                    # 모델의 랜덤성 조절. 낮을수록 일관적, 높을수록 창의적
    model_kwargs={
        "top_p":0.9,                    # 핵 샘플링. 0.1이면 상위 10% 확률만 고려
        "frequency_penalty":0.7,        # 같은 단어 반복 방지(양수일수록 강함)
        "presence_penalty":0.6,         # 새로운 주제로 유도(양자일수록 강함)
    },
    max_tokens = 2000 
)
```

```python
prompt = "시간 여행자는 조선을 방문하는 5문장의 짧은 소설을 써줘"
# 모델 객체에 전달할 프롬프트

response = creative_llm.invoke(prompt)
# 형성된 모델 객체에 프롬프트 전달

print(response.content)
# 프롬프트에 대한 답변이 담긴 content만 출력
```

### 정확한 답변용(Fact-based)

```python
from langchain_groq import ChatGroq

# 1. 정확한 답변용(Fact-based Model)
# - temperature 낮음(0.0) : 안정적이고 정확한 답변을 받음
Fact_llm = ChatGroq(
    model = "openai/gpt-oss-120b",      # 모델을 지정
    temperature=0.1,                    # 모델의 랜덤성 조절. 낮을수록 일관적, 높을수록 창의적
    model_kwargs={
        "top_p":1.0,                    # 핵 샘플링. 0.1이면 상위 10% 확률만 고려
        "frequency_penalty":0,          # 같은 단어 반복 방지(양수일수록 강함)
        "presence_penalty":0,           # 새로운 주제로 유도(양자일수록 강함)
    },
    max_tokens = 2000 
)
```

```python
prompt = "지구에서 가장 깊은 바다는 어디야?"
# 모델 객체에 전달할 프롬프트

response = creative_llm.invoke(prompt)
# 형성된 모델 객체에 프롬프트 전달

print(response.content)
# 프롬프트에 대한 답변이 담긴 content만 출력
```
