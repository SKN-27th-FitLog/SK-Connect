# Ollama Models

### 창의적 모델

```python
from langchain_ollama.chat_models import ChatOllama

creative_llm = ChatOllama(
    model = "gemma3:4b",     # 다운로드된 모델 명
    temperature = 0.9,       # 무작위성 높여 창의적인 결과 생성
    top_p = 0.9,             # 확률 분포 상위 90% 토큰만 사용(다양한 유지)
    num_predict=512,         # 충분한 길이의 텍스트 생성 허용
    keep_alive = "10m"       # 모델을 10분간 메모리에 유지(반복 생성 시 빠름)
)
```

```python
prompt = "시간 여행자가 조선을 방문하는 5문장의 짧은 소설을 써줘"

response = creative_llm.invoke(prompt)
# 모델 객체 형성 그리고 프롬프트 제공

print(response.content)
# 모델 객체에 전달한 프롬프트에 대한 답변의 내용을 출력
```

### 정확한 답변용
