# PromptTemplate
> PromptTemplate -> 기본 틀

## 다중 변주 프롬프트
#### 예제

```python
from langchain_core.prompts import PromptTemplate

# 주제, 대상, 톤을 정하고 추가적인 명령을 작성한 프롬프트
template = """ 
주제: {topic}
대상: {audience}
톤: {tone}

위 조건에 맞게 한국어로 간단하게 작성해줘
"""
```

```python
multi_var_prompt = PromptTemplate.from_template(
    template = template
)
# topic, audience, tone 변수를 포함한 프롬프트 템플릿 객체 생성
```

```python
multi_var_prompt.input_variables
# 들어가야할 변수를 확인
# 템플릿에 채워 넣어야 하는 변수 이름 목록 확인
```

```python
print(multi_var_prompt.template)
# multi_var_prompt에 넣은 template을 출력
```

```python
# format을 넣은 프롬프트 만들기
formatted_prompt2 = multi_var_prompt.format(
    topic="파이썬이란?",
    audience="일반인",
    tone="친근하고 이해하기 쉬운"
)

print("=== 다중 변수 프롬프트 ===")
print(formatted_prompt2)
# 템플릿 변수 자리에 실제 값을 넣어 완성된 프롬프트 문자열 생성
```

### 모델 불러오기(Ollama)

```python
from langchain_ollama import ChatOllama

model = ChatOllama(
    model = "gemma4:e4b",   # 모델 지정(호출)
    temperature = 0.1,      # 낮을수록 랜덤성이 줄어들어 더 일관되고 안정적인 답변 생성
    top_p = 1.0,            # 토큰 선택 시 누적 확률 범위를 설정; 1.0이면 확률 분포 전체를 사용
    num_predict = 3000,      # 새로 생성할 최대 토큰 수 제한
    keep_alive = "5m"       # 모델이 메모리에 5분 동안 유지해 재호출 시 빠르게 응답하도록 설정
)
```

### Langchain Expression Language
> 사용자 입력 -> 프롬프트 구성 -> LLM 호출 -> 결과 반환

#### PromptTemplate

```python
# chain 생성
# prompt와 model을 연결
chain = multi_var_prompt | model
```

```python
# 연결된 프롬프트와 모델에 대한 정보를 불러옴
# 생성된 객체 확인
chain
```

```python
# chain 호출
response = chain.invoke({
    "audience": "일반인",
    "tone": "친근하고 이해하기 쉬운",
    "topic": "파이썬이란?"
})
```

```python
response
```

```python
multi_var_prompt.template
# 모델에 대한 프롬프트 템플릿을 출력
```

```python
print(response.content)
# 모델에서 받은 답변에 대한 content만 출력
```
