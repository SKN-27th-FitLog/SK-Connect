```python
from langchain_community.document_loaders import CSVLoader

# CSV 파일을 LangChain Document 형태로 불러오기 위한 CSVLoader 설정
loader = CSVLoader(
    file_path=file_path,          # 불러올 CSV 파일 경로 (변수 file_path)
    encoding='utf-8',             # CSV 파일 인코딩 방식

    # Document의 본문(content)으로 사용할 컬럼 지정
    # 아래 컬럼 값들이 하나의 문자열로 합쳐져 page_content가 됨
    content_columns=['Open', 'High', 'Low', 'Close'],

    # Document의 source 정보로 사용할 컬럼 (예: 날짜)
    source_column="Date",

    # metadata로 포함할 컬럼들 (검색/필터 용도로 활용 가능)
    metadata_columns=['Volume', 'Change'],

    # CSV 파싱 설정
    csv_args={
        'delimiter': ',',         # CSV의 열 구분자 (기본 콤마)
    }
)
```

```python
docs2 = loader.load()

print(f"로드된 문서의 수:{len(docs2)}")
```

```python
for i, doc in enumerate(docs2[:2]):
    print("="*50)
    print(f"Document {i+1}:")
    print(f"   - 타입: {type(doc)}")
    print(f"   - page_content 타입: {type(doc.page_content)}")
    print(f"   - metadata 타입: {type(doc.metadata)}")
    print(f"   - metadata 내용: {doc.metadata}")
    print(f"   - 내용 길이: {len(doc.page_content)} 문자")
    print(f"   - 내용: \n{doc.page_content}")
    print()
```
