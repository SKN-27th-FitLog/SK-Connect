```python
# data 폴더에 있는 rag-keywords.txt 파일을 로드합니다
file_path = "./data/rag-keywords.txt"

# 파일 열기 및 읽기
with open(file_path, 'r', encoding='utf-8') as file:
    contents = file.read()
```

```python
# 내용 출력
print(f"[일부 앞부분 내용 확인]\n{contents[:300]}")
```

```python
print(f"저장된 파일의 위치정보: {file_path}")
```

```python
from langchain_community.document_loaders import TextLoader

# TextLoader는 텍스트 파일을 로드하는 Langchain의 기본 로더입니다
loader = TextLoader(file_path, encoding='utf-8')
```

```python
# load() 메서드를 호출하면 파일의 내용을 읽어서 Document 객체의 리스트로 반환합니다
documents = loader.load()
```

```python
print(f"로드된 파일의 수: {len(documents)}")
```

```python
for i, doc in enumerate(documents):
    print(f"Document {i+1}:")
    print(f"   - 타입: {type(doc)}")
    print(f"   - metadata 타입: {type(doc.metadata)}")
    print(f"   - metadata 내용: {doc.metadata}")
    print(f"   - page_content 타입: {type(doc.page_content)}")
    print(f"   - 내용 길이: {len(doc.page_content)} 문자")
    print(f"   - 내용: \n{doc.page_content[:100]}...")
```

# DirectoryLoader
> DirectoryLoader는 LangChain의 문서 로드 중 하나로, 지정한 디렉터리 안의 모든 파일을 자동으로 찾아서 각각을 Document로 변환한다. 

```python
from glob import glob

files = glob("./data/*.txt")

print(f"전체 txt 파일의 수: {len(files)}")
```

```python
files
```

```python
from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import TextLoader

loader = DirectoryLoader("./data", glob="*.txt", show_progress=True,
                         loader_cls=TextLoader, loader_kwargs={"encoding": "utf-8"})

documents = loader.load()

print(f"로드된 파일의 수: {len(documents)}")
```

```python
for i, doc in enumerate(documents):
    print("="*50)
    print(f"Document {i+1}:")
    print(f"   - 타입: {type(doc)}")
    print(f"   - metadata 타입: {type(doc.metadata)}")
    print(f"   - metadata 내용: {doc.metadata}")
    print(f"   - page_content 타입: {type(doc.page_content)}")
    print(f"   - 내용 길이: {len(doc.page_content)} 문자")
    print(f"   - 내용: \n{doc.page_content[:100]}...")
```
