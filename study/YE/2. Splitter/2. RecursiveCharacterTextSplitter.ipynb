# LangChain RAG 시스템 


## Loader 

```python
# LangChain에서 텍스트 파일을 불러오는 도구를 가져옴
from langchain_community.document_loaders import TextLoader

# 읽을 텍스트 파일 경로와 인코딩 방식을 지정해서 로더 생성
loader = TextLoader('./data/rag-keywords.txt', encoding='utf-8')

# 파일을 실제로 읽어서 docs 변수에 저장
docs = loader.load()
```

```python
print(f"문서의 수: {len(docs)}")
```

```python
# 문서의 메타데이터 확인 
docs[0].metadata
```

```python
# 문서 내용 확인 

print(docs[0].page_content[:300])
```

# RecursiveCharacterTextSplitter 
> 긴 텍스트를 작은 조각(청크)로 잘라주는 도구

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    # 청크 크기를 매우 작게 설정합니다. 예시를 위한 설정입니다.
    chunk_size=200,
    # 청크 간의 중복되는 문자 수를 설정합니다.
    chunk_overlap=20,
    # 문자열 길이를 계산하는 함수를 지정합니다.
    length_function=len,
    # 구분자로 정규식을 사용할지 여부를 설정합니다.
    is_separator_regex=False,
    # (옵션)한국어 문장 구분용 구분자 설정
    separators=[
        "\n\n",  # 문단
        "\n",    # 줄바꿈
        r"(?<=[.?!])\s+",  # 영어 문장
        r"(?<=[다요죠음함임니다])[\.\?!]?\s+",  # 한국어 종결 표현
        " ",      # 단어
        "",       # 최후의 수단: 글자 단위
    ]
)
```

```python
docs_with_splitter = text_splitter.split_documents(docs)

print(f"문서의 수: {len(docs_with_splitter)}")
```

### CharacterTextSplitter 와 다르게 최대 크기보다 큰 청크 생성될 수 없다. 

```python
for i, doc in enumerate(docs_with_splitter):
    if len(doc.page_content) > 200:
        print("="*50)
        print(f"{i+1}번째 200보다 큰 문장:\n{doc.page_content}")
```
